```python
import pandas as pd
import numpy as np
from datetime import datetime
import math

# Define regions of Cameroon
regions = [
    "Adamawa", "Centre", "East", "Far North", "Littoral",
    "North", "North-West", "South", "South-West", "West"
]

# Define 10 test districts per region
districts = [f"District{i}" for i in range(1, 11)]

# Generate monthly start dates from Jan 2024 to current month
months = pd.date_range("2024-01-01", datetime.today(), freq="MS")

rows = []
for month in months:
    # All days in the month
    days_in_month = pd.date_range(month, month + pd.offsets.MonthEnd(0), freq="D")
    # Pick 4 random days
    chosen_days = np.random.choice(days_in_month, size=4, replace=False)
    
    for date in chosen_days:
        date = pd.to_datetime(date)  # ensure Timestamp
        
        for region in regions:
            for district in districts:
                # Keep confirmed cases small (0–3 per district per day)
                confirmed = np.random.choice([0,0,1,2,3], p=[0.5,0.2,0.15,0.1,0.05])
                suspected = np.random.randint(0, 6)  # modest suspected cases
                
                # deaths capped so CFR ≤ 3%
                if confirmed > 0:
                    max_deaths = math.ceil(confirmed * 0.03)
                    deaths = np.random.randint(0, max_deaths + 1)
                else:
                    deaths = 0
                
                cfr = (deaths / confirmed * 100) if confirmed > 0 else 0
                rainfall = round(np.random.uniform(0, 20), 2)      # mm/day average
                temperature = round(np.random.uniform(20, 35), 1)  # °C
                humidity = round(np.random.uniform(50, 90), 1)     # %

                rows.append({
                    "date": date.strftime("%Y-%m-%d"),
                    "region": region,
                    "district": district,
                    "confirmed": confirmed,
                    "suspected": suspected,
                    "deaths": deaths,
                    "cfr": round(cfr, 2),
                    "rainfall": rainfall,
                    "temperature": temperature,
                    "humidity": humidity,
                    "created_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                })

# Build DataFrame
df = pd.DataFrame(rows)

# Save to CSV
df.to_csv("cameroon_cholera_reports_daily_districts_sample.csv", index=False)

print("CSV generated with", len(df), "rows")
print("Total confirmed cases:", df['confirmed'].sum())
```
