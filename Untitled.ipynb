```python
import pandas as pd

# Load your dataset
df = pd.read_csv("regional_cholera_data.csv")

def extract_district(location):
    if pd.isna(location):
        return None
    # Split by "::"
    parts = location.split("::")
    # Take the last non-empty part
    district = parts[-1].strip()
    return district

# Apply function to create new column
df["district"] = df["Location"].apply(extract_district)

# Quick check
print(df[["Location", "district"]].head())
df.to_csv("dataset_with_district.csv", index=False)
```

```python
import pandas as pd

# Load your dataset
df = pd.read_csv("dataset_with_district.csv")

# Function to extract district name
def extract_district(location):
    if pd.isna(location):
        return None
    parts = str(location).split("::")
    return parts[-1].strip()

# Create new column
df["district"] = df["Location"].apply(extract_district)

# List of region names in Cameroon
region_names = [
    "Adamawa","Adamaoua","Centre","Center","Yaounde","Yaoundé",
    "East","Est","Bertoua","Far North","Extreme Nord","Extrême-Nord","Maroua",
    "Littoral","Douala","North","Nord","Garoua","North West","Northwest",
    "Nord-Ouest","Bamenda","South","Sud","Ebolowa","South West","South-West","Penka-Michel",
    "Sud-Ouest","Buea","Fako","Meme","Manyu","Ndian","Kupe-Muanenguba",
    "Lebialem","West","Ouest","Bafoussam"
]

# Step 1: Remove rows where district contains "|" (multi-region strings)
df = df[~df["district"].str.contains(r"\|", na=False)]

# Step 2: Remove rows where district is a region name (case-insensitive)
df = df[~df["district"].str.strip().str.casefold().isin([r.casefold() for r in region_names])]

# Step 3: Handle duplicates
mask_health = df["district"].str.endswith("Health District", na=False)
health_districts = df[mask_health]["district"].str.replace(" Health District", "", regex=False)

# Drop duplicates where plain name exists alongside "Health District"
df = df[~((~mask_health) & (df["district"].isin(health_districts)))]

# Step 4: Keep only specified columns
df = df[["TL", "TR", "sCh", "cCh", "deaths", "region_en", "district"]]

# Save to new CSV
df.to_csv("dataset_with_only_district.csv", index=False)

print("Cleaned file saved as dataset_with_only_district.csv")
```
