```python
import pandas as pd
df = pd.read_csv("FINAL regional_cholera_environment_dataset.csv")
```

```python
import geopandas as gpd
dfg = gpd.read_file("cmr_admin_boundaries/cmr_admin2.geojson")
```

```python
print(df["Region"].sort_values().unique()[:30])
print(dfg["adm1_name"].sort_values().unique()[:30])
```

```python
df["Region"] = df["Region"].replace({
    "Adamaoua": "Adamawa",
    "Far North": "Far-North"
})
```

```python
import geopandas as gpd
import pandas as pd

# Dissolve ADM2 boundaries into ADM1 regions
regions_geo = dfg.dissolve(by="adm1_name", as_index=False)

# Aggregate cholera cases by region
region_cases = (
    df.groupby("Region")["deaths"]
    .sum()
    .reset_index()
)

# Merge map boundaries with cholera cases
region_map = regions_geo.merge(
    region_cases,
    left_on="adm1_name",
    right_on="Region",
    how="left"
)

region_map["deaths"] = region_map["deaths"].fillna(0)

print(region_map[["adm1_name", "Region", "deaths"]])
```

```python
df["reporting_date"] = pd.to_datetime(df["reporting_date"], errors="coerce")

for col in ["sCh", "cCh", "deaths"]:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)
```

```python
df["Region"] = df["Region"].replace({
    "Adamaoua": "Adamawa",
    "Far North": "Far-North"
})
```

```python
start_date = "2018-01-01"
end_date = "2018-12-31"

df_period = df[
    (df["reporting_date"] >= start_date) &
    (df["reporting_date"] <= end_date)
]
```

```python
region_stats = (
    df_period.groupby("Region")[["sCh", "cCh", "deaths"]]
    .sum()
    .reset_index()
)
```

```python
region_map = regions_geo.merge(
    region_stats,
    left_on="adm1_name",
    right_on="Region",
    how="left"
)

region_map[["sCh", "cCh", "deaths"]] = region_map[["sCh", "cCh", "deaths"]].fillna(0)

print(region_map[["adm1_name", "sCh", "cCh", "deaths"]])
```

```python
import plotly.express as px
fig = px.choropleth_mapbox(
    region_map,
    geojson=region_map.geometry,
    locations=region_map.index,
    color="cCh",
    hover_name="adm1_name",
    hover_data={
        "sCh": True,
        "cCh": True,
        "deaths": True
    },
    mapbox_style="carto-positron",
    center={"lat": 5.5, "lon": 12.5},
    zoom=5,
    opacity=0.7,
    title="Cholera Suspected Cases Hotspot Map"
)

fig.update_layout(margin={"r":0,"t":10,"l":0,"b":0})
fig.show()
```
