```python
import pandas as pd

# Load datasets
cholera = pd.read_csv("FINAL country_level_data_final.csv")         
pop_density = pd.read_csv("cmr_DENSITY.csv")    
water_sanitation = pd.read_csv("water&sanitation/washdash.csv")  

# Prepare cholera data
cholera['reporting_date'] = pd.to_datetime(cholera['reporting_date'])
cholera['Year'] = cholera['reporting_date'].dt.year

# Aggregate cholera data to yearly totals
yearly_cholera = cholera.groupby('Year').agg({
    'sCh':'sum', 
    'cCh':'sum',  
    'CFR':'mean'  
}).reset_index()


merged = yearly_cholera.merge(pop_density, on='Year', how='left')
merged = merged.merge(water_sanitation, on='Year', how='left')

merged.to_csv("merged_country_level_cholera_predictors.csv", index=False)


print(merged.head())
```

```python

df = pd.read_csv("merged_country_level_cholera_predictors.csv")         
df = df.drop(columns=['ISO3','Country'])
pivoted = df.pivot_table(
    index='Year',
    columns=['Service Type','Service level'],
    values='Coverage',
    aggfunc='mean'
).reset_index()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

final = df[['Year','cases','Population_Density']].drop_duplicates()
final = final.merge(pivoted, on='Year', how='left')

final = final.drop(columns=['Year'])

corr = final.corr(numeric_only=True)['cases'].sort_values(ascending=False)
filtered_corr = corr[(corr != 1.0) & (corr != -1.0) | (corr.index == 'cases')]
print(filtered_corr)
```

```python
corr_matrix = final.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, cmap='coolwarm', annot=True, fmt=".2f")
plt.title("Correlation Heatmap: Cholera Cases, Population Density, Water & Sanitation")
plt.show()
```
